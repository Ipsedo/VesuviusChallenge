```python
import torch as th
from vesuvius_challenge.networks.model import TrfAutoEncoder
```

```python
SIZES = (256, 256, 64)

BATCH_SIZE = 4
```

```python
x = th.randn(BATCH_SIZE, 1, *SIZES, device="cuda")
tgt_loss = th.rand(BATCH_SIZE, 1, *SIZES[:2], device="cuda")
tgt = tgt_loss.unsqueeze(-1).repeat(1, 1, 1, 1, SIZES[-1])
```

```python
trf = TrfAutoEncoder(
    [(1, 8), (8, 16), (16, 32), (32, 64)],
    8,
    64,
    8
).cuda()

loss_fn = th.nn.MSELoss()

optim = th.optim.Adam(trf.parameters(), lr=1e-4)
```

```python
out = trf(x, tgt)
```

```python
out.size()
```

```python
loss = loss_fn(out, tgt_loss)
```

```python
loss.backward()
```

```python
optim.step()
```
